# TP Plotly — Gapminder

*Objectif : s'initier Plotly pas à pas avec un dataset prétraité et intuitif.*


- On charge les données quand on en a besoin.
- On code par petits blocs.
- On visualise et on joue avec l'interactivité.

**Pré-requis** : Python, Jupyter, `plotly` installé.

## 0) Imports

````python
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
from plotly.subplots import make_subplots
````

## 1) Charger le dataset Gapminder

````python
# Gapminder est intégré à plotly.express
df = px.data.gapminder()
df.head()
````

## 2) Aperçu rapide des données

````python
# Dimensions et colonnes
df.shape, df.dtypes
````

````python
# Valeurs manquantes par colonne
df.isna().sum()
````

## 3) Scatter de base : PIB/hab vs espérance de vie
**À faire** : tracer un nuage de points `gdpPercap` vs `lifeExp` pour l'année 2007.

````python
df_2007 = df[df['year'] == 2007]
fig = px.scatter(
    df_2007,
    x='gdpPercap', y='lifeExp', color='continent',
    hover_name='country',
    title='PIB/hab vs Espérance de vie (2007)'
)
fig.show()
````

## 4) Bubble chart : ajouter la taille pour la population
**À faire** : réutiliser la figure précédente en ajoutant `size='pop'` et activer une échelle de taille raisonnable.

````python
fig = px.scatter(
    df_2007,
    x='gdpPercap', y='lifeExp', color='continent', size='pop',
    size_max=60, hover_name='country',
    title='PIB/hab vs Espérance de vie — Taille = population (2007)'
)
fig.show()
````

## 5) Facettes : séparer par continent
**À faire** : afficher une facette par continent (colonnes) et ajouter une échelle log sur l’axe x.

````python
fig = px.scatter(
    df_2007,
    x='gdpPercap', y='lifeExp', color='continent', size='pop',
    facet_col='continent', facet_col_wrap=3,
    size_max=50, hover_name='country',
    title='2007 — Facettes par continent'
)
fig.update_xaxes(type='log')
fig.for_each_annotation(lambda a: a.update(text=a.text.split('=')[-1]))
fig.show()
````

## 6) Animation dans le temps
**À faire** : animer le nuage de points de 1952 à 2007 (`animation_frame='year'`).

````python
fig = px.scatter(
    df,
    x='gdpPercap', y='lifeExp', color='continent', size='pop',
    hover_name='country',
    animation_frame='year', animation_group='country',
    log_x=True,   # 👈 gère correctement l’échelle log
    range_x=[100, 120_000],   # bornes cohérentes
    range_y=[20, 90],
    size_max=50,
    title='1952 → 2007 — PIB/hab vs Espérance de vie (animation)'
)
fig.show()
````

## 7) Courbes temporelles pour un sous-ensemble de pays
**À faire** : afficher l’évolution de `lifeExp` pour 5 pays choisis, avec une **légende cliquable** pour isoler un pays.
Astuce : sélectionner
```python
pays = ['France','Germany','Japan','Brazil','South Africa']
```


````python
pays = ['France','Germany','Japan','Brazil','South Africa']
dsel = df[df['country'].isin(pays)]
fig = px.line(
    dsel, x='year', y='lifeExp', color='country',
    title='Évolution de l\'espérance de vie pour quelques pays'
)
fig.update_traces(mode='lines+markers')
fig.update_layout(hovermode='x unified')
fig.show()
````

## 8) Carte choroplèthe : espérance de vie par pays (2007)
**À faire** : créer une carte du monde colorée par `lifeExp` en 2007, avec un slider d’année en bonus.

````python
fig = px.choropleth(
    df, locations='iso_alpha', color='lifeExp', hover_name='country',
    animation_frame='year', color_continuous_scale='Viridis',
    projection='natural earth',
    title='Espérance de vie par pays (1952–2007)'
)
fig.show()
````

## 9) Exporter une figure en HTML autonome
**À faire** : sauvegarder une figure en `.html` pour la partager dans un navigateur (fichier autonome).

````python
fig = px.scatter(df_2007, x='gdpPercap', y='lifeExp', color='continent', size='pop', hover_name='country',
                 title='2007 — Figure exportée en HTML')
fig.update_xaxes(type='log')
fig.write_html('plotly_gapminder_2007.html', include_plotlyjs=True)
print('Fichier écrit : plotly_gapminder_2007.html')
````
