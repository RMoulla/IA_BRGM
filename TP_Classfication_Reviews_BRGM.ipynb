# **Travaux pratiques : classification de texte (analyse de sentiment) avec du machine learning**

```python
import pandas as pd

dataset = pd.read_csv("Reviews.csv")
dataset.head()
```

```python
# Transformer la colonne recommended en variable numérique binaire avec astype(int)

dataset["recommended"] = dataset["recommended"].astype(int)
dataset.head(20)
```

```python
# Vectoriser la colonne text avec CountVectorizer
from sklearn.feature_extraction.text import CountVectorizer

# Initialiser le CountVectorizer
vectorizer = CountVectorizer()

# Transformer la colonne text en matrice de termes-documents
X = vectorizer.fit_transform(dataset["text"])

# Afficher la taille de la matrice obtenue
print(f"Taille de la matrice (documents, vocabulaire) : {X.shape}")
```

```python
# Afficher le premier commentaire vectorisé avec numpy en entier
import numpy as np

# Convertir la première ligne de la matrice creuse en un tableau numpy dense
first_vector_dense = X[0].toarray()

# Configurer numpy pour afficher tout le tableau sans troncature
with np.printoptions(threshold=np.inf):
    print(first_vector_dense)
```

## Modélisation avec une régression logistique

```python
# Partitionner les données en train et test

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, dataset["recommended"], test_size=0.2, random_state=42)
```

```python
# Entraîner un modèle de régression logistique
from sklearn.linear_model import LogisticRegression

# Initialiser le modèle de régression logistique
model = LogisticRegression()

# Entraîner le modèle
model.fit(X_train, y_train)

print("Modèle de régression logistique entraîné avec succès !")
```

```python
# Évaluer le modèle sur l'ensemble de test
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Prédire les classes pour l'ensemble de test
y_pred = model.predict(X_test)

# Calculer la précision
accuracy = accuracy_score(y_test, y_pred)
print(f"Précision du modèle (Accuracy) : {accuracy:.2%}\n")

# Afficher le rapport de classification détaillé
print("Rapport de classification :")
print(classification_report(y_test, y_pred))
```

```python
# Récupérer les dix mots qui poussent le plus vers un sentiment positif
import pandas as pd

# Récupérer les noms des fonctionnalités (mots) à partir du vectoriseur
feature_names = vectorizer.get_feature_names_out()

# Récupérer les coefficients du modèle de régression logistique
coefficients = model.coef_[0]

# Créer un DataFrame pour associer les mots à leurs coefficients
coef_df = pd.DataFrame({
    'Mot': feature_names,
    'Coefficient': coefficients
})

# Trier par ordre décroissant pour obtenir les coefficients les plus élevés (positifs)
top_10_positive = coef_df.sort_values(by='Coefficient', ascending=False).head(10)

# Afficher les résultats
display(top_10_positive)
```

```python
# Récupérer les dix mots qui poussent le plus vers un sentiment négatifs
import pandas as pd

# Récupérer les noms des fonctionnalités (mots) à partir du vectoriseur
feature_names = vectorizer.get_feature_names_out()

# Récupérer les coefficients du modèle de régression logistique
coefficients = model.coef_[0]

# Créer un DataFrame pour associer les mots à leurs coefficients
coef_df = pd.DataFrame({
    'Mot': feature_names,
    'Coefficient': coefficients
})

# Trier par ordre décroissant pour obtenir les coefficients les plus bas (négatifs)
top_10_negative = coef_df.sort_values(by='Coefficient').head(10)

# Afficher les résultats
display(top_10_negative)
```
